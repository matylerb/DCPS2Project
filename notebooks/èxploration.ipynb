```python
import pandas as pd
import numpy as np
from ydata_profiling import ProfileReport
import matplotlib.pyplot as plt
```

```python
def explore_data():
    df = pd.read_csv("../data/processed/cleaned_data.csv")
    df.info()
    df.describe()
    df.head()
```

```python
import glob
import os

csv_files = glob.glob("../data/processed/*.csv")

#eda tool that generates a report for each csv file in the processed data directory
for filepath in csv_files:
    df = pd.read_csv(filepath)
    filename = os.path.splitext(os.path.basename(filepath))[0]
    report = ProfileReport(df, title=f"EDA Report: {filename}")
    report.to_file(f"../outputs/reports/{filename}_report.html")
    print(f"Report saved for {filename}")
```
